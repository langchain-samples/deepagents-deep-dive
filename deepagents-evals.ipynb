# Evaluating a Deep Agent

A single agent run can look convincing and still hide a brittle system. Did it follow the traveler's constraints, or did this prompt just happen to work? Did the specialist actually leave behind the artifact the supervisor promised?

We'll answer those questions with a small **Kyoto travel concierge**. A supervisor delegates itinerary design to a local-guide subagent, the guide consults a fixed activity catalog and writes `/itinerary.md`, and the supervisor reads that artifact and returns it to the user.

Then we'll turn that one demonstration into an offline LangSmith experiment with exactly two scores:

- **Itinerary quality** — is the user-facing plan correctly formatted, grounded in the catalog, practical, and faithful to every constraint?
- **Artifact handoff** — does the backing file exist, and did the supervisor return its contents faithfully?

The catalog is deliberately fixed rather than live travel data. That keeps the lesson reproducible: when a score changes, the agent changed — the world did not.

```python
import json
from typing import Annotated, TypedDict

from dotenv import load_dotenv
load_dotenv(override=True)

from IPython.display import Code, Markdown, display
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.tools import tool
from langsmith import Client
from deepagents import create_deep_agent

model = "claude-haiku-4-5-20251001"
client = Client()
```

# The scenario: one city, three kinds of traveler

Our concierge does not search the web. Its local-guide subagent gets one tool, `lookup_activities`, backed by this compact catalog:

| Activity | Duration | Cost | Useful constraints |
|---|---:|---:|---|
| Kennin-ji Temple | 60 min | ¥600 | temple, culture, quiet, mostly indoors |
| Machiya Tea Workshop | 60 min | ¥3,500 | tea, culture, indoors |
| Kyoto International Manga Museum | 90 min | ¥1,200 | museum, family-friendly, indoors |
| Nishiki Market Tasting | 60 min | ¥2,000 | food, lunch, covered, crowded |
| Shosei-en Garden | 75 min | ¥500 | garden, quiet, outdoors |
| Furoshiki Workshop | 60 min | ¥2,500 | traditional craft, hands-on, indoors |
| Kyoto Railway Museum | 120 min | ¥1,500 | museum, family-friendly, accessible, indoors |

The data is small enough to inspect at a glance, but rich enough to create real tradeoffs around time, cost, weather, and interests.


```python
KYOTO_ACTIVITIES = [
    {
        "name": "Kennin-ji Temple",
        "duration_minutes": 60,
        "cost_yen": 600,
        "tags": ["temple", "culture", "quiet", "mostly indoors"],
    },
    {
        "name": "Machiya Tea Workshop",
        "duration_minutes": 60,
        "cost_yen": 3500,
        "tags": ["tea", "culture", "indoors"],
    },
    {
        "name": "Kyoto International Manga Museum",
        "duration_minutes": 90,
        "cost_yen": 1200,
        "tags": ["museum", "family-friendly", "indoors"],
    },
    {
        "name": "Nishiki Market Tasting",
        "duration_minutes": 60,
        "cost_yen": 2000,
        "tags": ["food", "lunch", "covered", "crowded"],
    },
    {
        "name": "Shosei-en Garden",
        "duration_minutes": 75,
        "cost_yen": 500,
        "tags": ["garden", "quiet", "outdoors"],
    },
    {
        "name": "Furoshiki Workshop",
        "duration_minutes": 60,
        "cost_yen": 2500,
        "tags": ["traditional craft", "hands-on", "indoors"],
    },
    {
        "name": "Kyoto Railway Museum",
        "duration_minutes": 120,
        "cost_yen": 1500,
        "tags": ["museum", "family-friendly", "accessible", "indoors"],
    },
]


@tool
def lookup_activities() -> list[dict]:
    """Return the approved Kyoto activity catalog with durations, costs, and tags."""
    return KYOTO_ACTIVITIES
```

# A specialist with one job

Itinerary design is a clean subagent boundary. The supervisor only needs to recognize a travel-planning request and delegate it. The local guide owns the catalog, the formatting contract, and the actual file write.

The user asks for an itinerary — not a filename. `/itinerary.md` is an internal contract between the supervisor and specialist: the guide writes it, the supervisor reads it, and the user simply gets the result.

The output is deliberately vertical rather than tabular so it stays readable in a narrow notebook. Each stop carries its own time, cost, and rationale, followed by one compact trip summary.


```python
ITINERARY_PATH = "/itinerary.md"

ITINERARY_FORMAT = """# Kyoto Itinerary

1. **HH:MM–HH:MM — Exact catalog activity name** — ¥0
   - Why: One concise reason

2. **HH:MM–HH:MM — Exact catalog activity name** — ¥0
   - Why: One concise reason

3. **HH:MM–HH:MM — Exact catalog activity name** — ¥0
   - Why: One concise reason

## Trip summary
- **Duration:** ...
- **Estimated cost:** ...
- **Constraints met:** ...
"""

local_guide = {
    "name": "local-guide",
    "description": (
        "Design a constrained Kyoto itinerary using the approved activity catalog, "
        f"write it to {ITINERARY_PATH}, and return only that path."
    ),
    "system_prompt": (
        "You are a Kyoto local-guide specialist. Always call lookup_activities before planning. "
        "Use only activities and facts returned by that tool. Create exactly three stops whose "
        "time ranges match the catalog durations, leaving gaps when useful for transit. Honor every "
        f"traveler constraint. Write {ITINERARY_PATH} using this exact template and return only the path.\n\n"
        + ITINERARY_FORMAT
    ),
    "tools": [lookup_activities],
}

supervisor = create_deep_agent(
    model=model,
    system_prompt=(
        "You are a Kyoto travel concierge. Delegate every itinerary request to the local-guide "
        f"subagent. After it returns, read {ITINERARY_PATH} yourself and return its complete contents "
        "verbatim as your final response. Do not mention the internal file path or add a preamble."
    ),
    subagents=[local_guide],
)
```

# The target: what LangSmith runs

In an offline evaluation, the **target** is the system under test. It is an ordinary function that LangSmith calls once for every dataset example.

<div style="font-size:17px; text-align:center; padding:12px; border:1px solid #d0d7de; border-radius:6px;">
dataset inputs &nbsp;→&nbsp; <code>target(inputs)</code> &nbsp;→&nbsp; target outputs &nbsp;→&nbsp; evaluators
</div>

Our target has three jobs:

1. Read the traveler's <code>query</code> from the dataset input.
2. Invoke the supervisor exactly as the application would.
3. Adapt the resulting deep-agent state into a stable <code>{response, files}</code> shape for the evaluators.

The target does **not** grade anything and never sees the reference rubric. It only runs the application and returns what happened.

Before writing evaluators, we run this target once and inspect its real output. Deep-agent state contains both messages and the shared filesystem; the small explicit return shape makes those two layers easy to grade.


```python
def run_travel_agent(inputs: dict) -> dict:
    result = supervisor.invoke(
        {"messages": [{"role": "user", "content": inputs["query"]}]}
    )

    files = {
        path: file_data["content"]
        for path, file_data in (result.get("files") or {}).items()
        if file_data.get("encoding", "utf-8") == "utf-8"
    }
    return {
        "response": result["messages"][-1].text, # Orchestrator's final response to user
        "files": files, # All files in filesystem
    }
```

```python
anchor_request = (
    "Plan a four-hour rainy afternoon in Kyoto for under ¥7,000. "
    "Include one temple and one tea experience, and minimize outdoor time."
)

demo = run_travel_agent({"query": anchor_request})

assert set(demo) == {"response", "files"}
artifact = demo["files"].get(ITINERARY_PATH)
assert isinstance(artifact, str) and artifact.strip()

display(Markdown(
    "## Target output\n\n"
    "The target returned two fields. They are shown separately below."
))

display(Markdown(
    "### `response`\n\n"
    "*The user-facing Markdown returned by the supervisor.*"
))
display(Markdown(demo["response"]))

display(Markdown(
    "### `files`\n\n"
    "*The internal filesystem artifacts returned with the agent state.*"
))
for path, content in demo["files"].items():
    display(Markdown(f"#### `{path}`"))
    display(Code(content, language="markdown"))
```

Everything under **Target output** came from the dictionary returned by `run_travel_agent`. The labels are notebook display scaffolding, not part of the agent's response:

- `response` is rendered as Markdown because it is what the user sees.
- `files` is shown as raw Markdown source so the internal artifacts look visibly different.

In this run the two contents should match: the local guide wrote the itinerary, and the supervisor read and returned it verbatim. That exact handoff is what the deterministic evaluator will check.

This demonstration still does **not** establish reliability. To do that, we need several traveler profiles and explicit definitions of success.


# Turn expectations into a dataset

Each example has one input, `query`, and one reference output, `rubric`:

- `query` is the traveler request passed to the target.
- `rubric` describes the example-specific conditions for success. It is not an expected answer to copy.

The artifact path does not vary by example, so it does not belong in the references. It is a system-wide invariant captured once by `ITINERARY_PATH`.

The dataset name is versioned. Rerunning the notebook safely reuses an existing complete dataset instead of deleting remote data.


```python
DATASET_NAME = "deepagents-kyoto-itinerary-evals-v2"

examples = [
    {
        "inputs": {"query": anchor_request},
        "outputs": {
            "rubric": (
                "The itinerary lasts no more than four hours, costs at most ¥7,000, includes "
                "Kennin-ji Temple and Machiya Tea Workshop, and minimizes outdoor time for rain."
            ),
        },
    },
    {
        "inputs": {
            "query": (
                "Plan a five-hour family afternoon in Kyoto for one adult and a ten-year-old, "
                "under ¥7,500. Include the Kyoto Railway Museum, a lunch stop, and one hands-on "
                "activity."
            )
        },
        "outputs": {
            "rubric": (
                "The itinerary lasts no more than five hours, costs at most ¥7,500, includes "
                "Kyoto Railway Museum, Nishiki Market Tasting for lunch, and Furoshiki Workshop "
                "as a hands-on activity."
            ),
        },
    },
    {
        "inputs": {
            "query": (
                "Plan a quiet four-hour solo morning in Kyoto for under ¥5,000. Include a garden "
                "and a traditional craft, and avoid crowded activities."
            )
        },
        "outputs": {
            "rubric": (
                "The itinerary lasts no more than four hours, costs at most ¥5,000, includes "
                "Shosei-en Garden and Furoshiki Workshop, avoids Nishiki Market Tasting, and uses "
                "a quiet third stop."
            ),
        },
    },
]

if client.has_dataset(dataset_name=DATASET_NAME):
    dataset = client.read_dataset(dataset_name=DATASET_NAME)
else:
    dataset = client.create_dataset(
        DATASET_NAME,
        description="Constraint-following and artifact-handoff evals for a Kyoto travel concierge.",
    )

existing_examples = list(client.list_examples(dataset_id=dataset.id))
if not existing_examples:
    client.create_examples(dataset_id=dataset.id, examples=examples)
    print(f"Created {len(examples)} examples.")
else:
    remote_examples = [
        {"inputs": example.inputs, "outputs": example.outputs}
        for example in existing_examples
    ]
    by_query = lambda example: example["inputs"]["query"]
    if sorted(remote_examples, key=by_query) != sorted(examples, key=by_query):
        raise RuntimeError(
            f"{DATASET_NAME!r} already exists with different examples. "
            "Use a new versioned dataset name rather than overwriting remote data."
        )
    print(f"Reusing {DATASET_NAME!r} with {len(existing_examples)} examples.")

print(f"Dataset: {dataset.url}")
```

One evaluator returns one metric. Keeping the questions separate makes failures actionable: itinerary quality measures the plan; artifact handoff measures the architecture that delivered it.


## 1. Itinerary quality: a strict structured judge

The reference is a rubric, not a golden response. Many itineraries can be good, but every passing user-facing response must satisfy the same exact contract.

`ITINERARY_FORMAT` is that contract. The local guide receives it as its output instruction, and the judge receives the same value as its grading reference. Defining the product requirement once prevents the two descriptions from drifting.

The judge returns four criterion-level booleans plus an explanation. We collapse them into one binary LangSmith score: all four must pass.


```python
class ItineraryGrade(TypedDict):
    format_ok: Annotated[
        bool,
        "The response follows the shared ITINERARY_FORMAT contract with every placeholder filled.",
    ]
    catalog_grounded: Annotated[
        bool,
        "Every stop name, duration, cost, and factual claim agrees with the supplied catalog.",
    ]
    schedule_and_math_ok: Annotated[
        bool,
        "Times are chronological and non-overlapping; durations and costs are correct and within limits.",
    ]
    constraints_ok: Annotated[
        bool,
        "Every request and rubric constraint is satisfied and named under Constraints met.",
    ]
    reasoning: Annotated[str, "A concise explanation naming every failed criterion."]


QUALITY_CHECKS = (
    "format_ok",
    "catalog_grounded",
    "schedule_and_math_ok",
    "constraints_ok",
)


def quality_score(grade: ItineraryGrade) -> int:
    return int(all(grade[check] for check in QUALITY_CHECKS))


judge_prompt = f"""You are a strict evaluator of a user-facing Kyoto itinerary.

The response must follow this output contract:

<itinerary_format>
{ITINERARY_FORMAT}
</itinerary_format>

Set format_ok=true only when the response follows that contract with every placeholder replaced by
a real value, exactly three numbered stops, one Why bullet per stop, and no missing sections.

Set catalog_grounded=true only when every activity name is copied exactly from the catalog and its
listed duration, cost, tags, and descriptive claims agree with the catalog.

Set schedule_and_math_ok=true only when time ranges are chronological and non-overlapping, each
activity's time range matches its catalog duration, the stated cost equals the sum of all three
activities, and the elapsed itinerary duration and cost stay within the user's limits. Gaps for
transit are allowed and count toward elapsed duration.

Set constraints_ok=true only when the itinerary satisfies both the user's request and the success
rubric, and the Constraints met line explicitly names every requested constraint.

Be strict: all four checks must independently earn true. Do not excuse missing structure because the
prose sounds helpful.
"""

judge = init_chat_model(model, temperature=0).with_structured_output(ItineraryGrade)


def itinerary_quality(inputs, outputs, reference_outputs):
    itinerary = outputs.get("response", "")
    if not isinstance(itinerary, str) or not itinerary.strip():
        return {
            "key": "itinerary_quality",
            "score": 0,
            "comment": "The supervisor returned no user-facing itinerary.",
        }

    grade = judge.invoke([
        SystemMessage(content=judge_prompt),
        HumanMessage(content=(
            f"User request:\n{inputs['query']}\n\n"
            f"Success rubric:\n{reference_outputs['rubric']}\n\n"
            f"Approved activity catalog:\n"
            f"{json.dumps(KYOTO_ACTIVITIES, ensure_ascii=False, indent=2)}\n\n"
            f"User-facing response:\n{itinerary}"
        )),
    ])

    checks = ", ".join(
        f"{name}={'pass' if grade[name] else 'fail'}" for name in QUALITY_CHECKS
    )
    return {
        "key": "itinerary_quality",
        "score": quality_score(grade),
        "comment": f"{checks}. {grade['reasoning']}",
    }
```

## 2. Artifact handoff: use exact matching for an exact invariant

The supervisor is instructed to return the file verbatim. That gives us a deterministic contract: `/itinerary.md` must exist, be non-empty, and match the user-facing response.

The evaluator normalizes only line endings and surrounding whitespace. Any difference in the actual itinerary content fails.


```python
def normalize_text(value: str) -> str:
    return value.replace("\r\n", "\n").strip()


def artifact_handoff(outputs):
    response = outputs.get("response")
    artifact = (outputs.get("files") or {}).get(ITINERARY_PATH)

    response_ok = isinstance(response, str) and bool(response.strip())
    artifact_ok = isinstance(artifact, str) and bool(artifact.strip())
    passed = (
        response_ok
        and artifact_ok
        and normalize_text(response) == normalize_text(artifact)
    )

    if passed:
        comment = f"The supervisor response matches the artifact at {ITINERARY_PATH}."
    elif not artifact_ok:
        comment = f"Missing or empty UTF-8 artifact at {ITINERARY_PATH}."
    elif not response_ok:
        comment = "The supervisor returned an empty response."
    else:
        comment = f"The supervisor response does not match the artifact at {ITINERARY_PATH}."

    return {
        "key": "artifact_handoff",
        "score": int(passed),
        "comment": comment,
    }
```

# Test the evaluators before spending experiment calls

The LLM judge itself is exercised by the experiment, but its score aggregation is ordinary code. The handoff evaluator is fully deterministic, so we test missing, mismatched, and normalized-equivalent content before running three agents and three judges.


```python
good_grade = {
    "format_ok": True,
    "catalog_grounded": True,
    "schedule_and_math_ok": True,
    "constraints_ok": True,
    "reasoning": "All checks pass.",
}
assert quality_score(good_grade) == 1
assert quality_score({**good_grade, "constraints_ok": False}) == 0

assert artifact_handoff(
    {"response": "# Kyoto Itinerary", "files": {}},
)["score"] == 0
assert artifact_handoff(
    {"response": "# Kyoto Itinerary", "files": {ITINERARY_PATH: "Different"}},
)["score"] == 0
assert artifact_handoff(
    {
        "response": "  # Kyoto Itinerary\nStop 1  ",
        "files": {ITINERARY_PATH: "# Kyoto Itinerary\r\nStop 1"},
    },
)["score"] == 1

print("Evaluator checks passed.")
```

# Run the experiment

LangSmith now applies the same target and the same two evaluators to every traveler profile. The result is a comparable experiment rather than three anecdotes.

`max_concurrency=2` keeps the demo moving without sending all model calls at once.


```python
results = client.evaluate(
    run_travel_agent,
    data=DATASET_NAME,
    evaluators=[itinerary_quality, artifact_handoff],
    experiment_prefix="kyoto-concierge",
    max_concurrency=2,
)

print(f"Experiment: {results.experiment_name}")
print(f"View results: {results.url}")
```
